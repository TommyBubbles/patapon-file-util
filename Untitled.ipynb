## function definition

```python
def c_type_to_field_metadata_type(c_type: str) -> str:
    mapping = {
        "char": "string",
        "int32": "signed_int",
        "uint32": "unsigned_int",
        "int16": "signed_int16",
        "uint16": "unsigned_int16",
        "int8": "signed_int8",
        "uint8": "unsigned_int8",
        "float": "float"
    }
    return mapping[c_type]


def c_type_to_python_type(c_type: str) -> str:
    mapping = {
        "char": "str",
        "int32": "int",
        "uint32": "int",
        "int16": "int",
        "uint16": "int",
        "int8": "int",
        "uint8": "int",
        "float": "float"
    }
    return mapping[c_type]


def get_default_value(c_type: str) -> str:
    mapping = {
        "char": "\"\"",
        "int32": "0",
        "uint32": "0",
        "int16": "0",
        "uint16": "0",
        "int8": "0",
        "uint8": "0",
        "float": "0.0"
    }
    return mapping[c_type]


def print_struct(struct, tabs:int=0):
    for j in struct.__fields__:
        if struct.__fields_types__[j].c_type == 'struct':
            print("\t" * tabs + j)
            print_struct(getattr(struct, j), tabs+1)
        else:
            print("\t" * tabs + j + " " + str(getattr(struct, j)))


def print_object_def(struct, level=0):
    if level == 0:
        print("(")
    for j in struct.__fields__:
        if struct.__fields_types__[j].c_type == 'struct':
            value = getattr(struct, j)
            if type(value) == list:
                for i in range(len(value)):
                    print_object_def(value[i], level+1)
            else:
                print_object_def(value, level+1)
        elif struct.__fields_types__[j].c_type == 'char':
            try:
                getattr(struct, j).decode()
                print(f'{getattr(struct, j)}.decode(),')
            except UnicodeDecodeError:
                print(f'{getattr(struct, j)}.decode("shift-jis"),')
        else:
            print(f'{getattr(struct, j)},')
    if level == 0:
        print(")")
```

```python
def field_string(struct, skip_levels=-1):
    _field_string(struct, skip_levels)


def _field_string(struct, skip_levels, level=0, index=0, postfix="") -> int:
    if skip_levels != -1 and skip_levels <= level:
        return index
    
    ordered_fields = sorted(struct.__fields_types__.items(), key=lambda x: x[1].offset)
    for name, info in ordered_fields:
        if info.c_type == "struct":
            value = getattr(struct, name)
            if type(value) == list:
                for i in range(len(value)):
                    new_index = _field_string(value[i], skip_levels, level+1, index, f"{postfix}_{i+1}")
                if new_index == index:
                    signature = f"{name}{postfix}"
                    print(f"    {signature}")
                index = new_index + 1
            else:
                new_index = _field_string(value, skip_levels, level+1, index, postfix)
                if new_index == index:
                    signature = f"{name}{postfix}"
                    print(f"    {signature}")
                index = new_index + 1
        else:
            if info.c_type != 'char' and info.vlen_ex > 1:
                signature = f"{name}{postfix}: list[{c_type_to_python_type(info.c_type)}]"
                default_def = f"default_factory=list[{c_type_to_python_type(info.c_type)}]"
            else:
                signature = f"{name}{postfix}: {c_type_to_python_type(info.c_type)}"
                default_def = f'default={get_default_value(info.c_type)}'
            metadata_def = 'metadata={"meta": FieldMetadata("' + c_type_to_field_metadata_type(info.c_type) + '", ' + str(index)
            if info.c_type == 'char':
                metadata_def += f', size={hex(info.vlen_ex)}'
            elif info.vlen_ex > 1:
                metadata_def += f', count={info.vlen_ex}'
            metadata_def += ')}'
            field_def = f'field({default_def}, {metadata_def})'
            index += 1
            print(f"    {signature} = {field_def}")
    return index
```
